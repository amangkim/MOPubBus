```python
import pandas as pd

train_ratio = 0.8
```

```python
normal = 'sorted_mapped_bus_data_AP1_normal.csv'
grandprix = 'sorted_mapped_bus_data_AP1_grandprix.csv'

normal = pd.read_csv(normal)
grandprix = pd.read_csv(grandprix)
```

```python
def split_by_ratio(df, train_ratio):
    split_index = int(len(df) * train_ratio)
    train_df = df.iloc[:split_index].reset_index(drop=True)
    test_df = df.iloc[split_index:].reset_index(drop=True)
    return train_df, test_df
```

```python
train_normal, test_normal = split_by_ratio(normal, train_ratio)
train_grandprix, test_grandprix = split_by_ratio(grandprix, train_ratio)

print("Normal:")
print("Train:", len(train_normal), "Test:", len(test_normal))

print("Grandprix:")
print("Train:", len(train_grandprix), "Test:", len(test_grandprix))
```

```python
train_normal.to_csv('AP1_normal_train.csv', index=False)
test_normal.to_csv('AP1_normal_test.csv', index=False)

train_grandprix.to_csv('AP1_grandprix_train.csv', index=False)
test_grandprix.to_csv('AP1_grandprix_test.csv', index=False)
```
